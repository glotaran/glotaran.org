# Going further 🚀

You have run three complete analyses — two on simulated data, one on a
real fluorescence measurement — and you know the cycle at the heart of
global and target analysis:

> **inspect the data → specify a model → estimate the parameters →
> validate and interpret**

In the literature this is the *scientific discovery cycle* of **model
specification → parameter estimation → model validation**: you propose
a model, fit it, inspect the residuals, and refine the model until it
describes the data with interpretable spectra and kinetics. Almost
everything else in pyglotaran is a variation on this cycle.

This notebook is a **map of where to go next**: the key concepts from
the literature, links to real worked case studies, and the papers
behind the method. There is no code to run here — just directions.

## Next techniques to learn

Each of these is a small step beyond what you have already done:

- **Target analysis** — instead of a convenient sequential
  `s1 → s2 → …` scheme, you describe the actual physical kinetic scheme
  (branching, equilibria, competing pathways) in a compartmental model
  and recover the true **species-associated spectra (SADS/SAS)**. The
  fluorescence study you just ran also has a *target* version.
- **Multiple datasets at once** — analyse several measurements
  simultaneously so they share parameters (for example, the same
  kinetics across different excitation wavelengths).
- **Spectral constraints and guidance** — bring in *a priori* spectral
  knowledge (known zeros, equalities, area penalties, or a reference
  spectrum) to make the parameters identifiable and stabilise a hard
  fit.
- **Damped oscillations (DOAS)** — model coherent oscillatory signals
  (vibrational wavepackets) on top of the decays, together with a
  parametric description of the **coherent artifact**.
