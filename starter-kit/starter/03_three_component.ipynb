# Going deeper: a three-component dataset 🔬🔬

This notebook runs the **same cycle as notebook 02** — inspect,
specify, estimate, validate — on a slightly richer dataset. The sample
now relaxes through **three states in sequence**
(s1 → s2 → s3 → ground) instead of two.

This is the whole point of global analysis: adding a state is a small
change to the kinetic scheme, and everything else — inspecting,
estimating, interpreting — stays exactly the same.

Run each cell with **Shift + Enter**, in order.

## Step 0: Load the tools we need

```python
from csv_tools import load_csv_dataset

from glotaran.io import save_result
from glotaran.optimization.optimize import optimize
from glotaran.project.scheme import Scheme

from pyglotaran_extras import plot_data_overview, plot_overview
```

## Step 1: Inspect the data

The data is in `03_three_component/data.csv`, in the same layout as
before (time down the first column, wavelength across the first row).

Look at the overview: compared with notebook 02 you should be able to
see an extra feature evolving in time — the signature of the third
state — telling you the model will need one more compartment.

```python
data = load_csv_dataset("03_three_component/data.csv")

plot_data_overview(data, linlog=True, linthresh=1);
```

## Step 2: Specify the model and starting parameters

- `03_three_component/model.yaml` — a kinetic scheme of three states
  decaying in sequence. Open it side by side with
  `02_two_component/model.yaml`: the only real difference is the extra
  state and one more rate.
- `03_three_component/parameters.yaml` — the starting guesses, now with
  three decay rates.

```python
scheme = Scheme(
    model="03_three_component/model.yaml",
    parameters="03_three_component/parameters.yaml",
    data={"dataset1": data},
    maximum_number_function_evaluations=99,
)

scheme.validate()
```

## Step 3: Estimate the parameters (optimize)

```python
result = optimize(scheme)
result
```

```python
result.optimized_parameters
```

## Step 4: Validate and interpret the results

There are now three concentration profiles and three evolution-
associated spectra, one per state. As before, first check the fit
describes the data (the residuals), then read off the science.
Everything else about the plot is the same as in notebook 02.

```python
plot_overview(result.data["dataset1"], linlog=True, linthresh=10);
```

## (Optional) Save the results

```python
save_result(
    result=result,
    result_path="03_three_component/results/result.yml",
    allow_overwrite=True,
)
print("Saved! Look in the 03_three_component/results folder on the left.")
```
